```python
# Nomor 1
# inverse citra
original = cv.imread('/content/drive/MyDrive/PCVK/Gambar/apel.jpg')

inverse_image = 255 - original
final_frame = cv.hconcat((original, inverse_image))
cv2_imshow(final_frame)
```

```python
# Nomor 2
# transformasi contrast

try:
    brightness = int(input('Masukkan tingkat kecerahan : '))
    contrast = float(input('Masukkan kontras : '))
except ValueError:
    print('Error, not a number')
    brightness = 0
    contrast = 1

original = cv.imread('/content/drive/MyDrive/PCVK/Gambar/houses.jpg')

contrast_image = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            contrast_image[y,x,c] = np.clip(
                contrast * int(original[y,x,c]) + brightness, 0, 255)

final_frame = cv.hconcat((original, contrast_image))
cv2_imshow(final_frame)
```

```python
# Nomor 3
# Transformasi Logarithmic Brightness

try:
    brightness = int(input('Masukkan nilai kecerahan: '))
except ValueError:
    print('Error, not a number')
    brightness = 1

original = cv.imread('/content/drive/MyDrive/PCVK/Gambar/houses.jpg')
log_image = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            log_image[y,x,c] = np.clip(
                brightness * np.log(1 + int(original[y,x,c])), 0, 255 )

final_frame = cv.hconcat((original, log_image))
cv2_imshow(final_frame)
```

```python
# Nomor 4
# Grayscale: Averaging, Lightness, dan Luminance

original = cv.imread('/content/drive/MyDrive/PCVK/Gambar/apel.jpg')
B, G, R = cv.split(original)

# 1. Averaging
averaging = (R.astype(float) + G.astype(float) + B.astype(float)) / 3
averaging = np.uint8(averaging)

# 2. Lightness
lightness = (np.maximum(np.maximum(R, G), B).astype(float) + np.minimum(np.minimum(R, G), B).astype(float)) / 2
lightness = np.uint8(lightness)

# 3. Luminance
luminance = (0.21 * R.astype(float) + 0.72 * G.astype(float) + 0.07 * B.astype(float))
luminance = np.uint8(luminance)

hasil_averaging = cv.hconcat((original, cv.cvtColor(averaging, cv.COLOR_GRAY2BGR)))
hasil_lightness = cv.hconcat((original, cv.cvtColor(lightness, cv.COLOR_GRAY2BGR)))
hasil_luminance = cv.hconcat((original, cv.cvtColor(luminance, cv.COLOR_GRAY2BGR)))
cv2_imshow(hasil_averaging)
cv2_imshow(hasil_lightness)
cv2_imshow(hasil_luminance)
```

```python
# Nomor 5
# Menampilkan warna merah dan mengubah warna lain menjadi grayscale

import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

original = cv.imread('/content/drive/MyDrive/PCVK/Gambar/apel.jpg')
hsv = cv.cvtColor(original, cv.COLOR_BGR2HSV)

lower_red1 = np.array([0, 100, 50])
upper_red1 = np.array([10, 255, 255])
lower_red2 = np.array([170, 100, 50])
upper_red2 = np.array([180, 255, 255])
mask1 = cv.inRange(hsv, lower_red1, upper_red1)
mask2 = cv.inRange(hsv, lower_red2, upper_red2)
mask = cv.bitwise_or(mask1, mask2)
gray = cv.cvtColor(original, cv.COLOR_BGR2GRAY)
gray = cv.cvtColor(gray, cv.COLOR_GRAY2BGR)

hasil = np.where(mask[:, :, np.newaxis] > 0, original, gray)
hasil = np.uint8(hasil)

final_frame = cv.hconcat((original, hasil))
cv2_imshow(final_frame)
```

```python
# Nomor 6
# Gamma Correction

import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
try:
    gamma = float(input('Masukkan nilai Gamma: '))
except ValueError:
    print('Error, not a number')
    gamma = 1

original = cv.imread('/content/drive/MyDrive/PCVK/Gambar/houses.jpg')
gamma_image = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        for c in range(original.shape[2]):
            gamma_image[y, x, c] = np.clip(255 * ((int(original[y, x, c]) / 255) ** (1 / gamma)),0,255)

final_frame = cv.hconcat((original, gamma_image))
cv2_imshow(final_frame)
```

```python
# Nomor 7
# Simulasi Image Depth

bit_depth = int(input('Masukkan bit depth tujuan (1-8): '))
bit_depth_awal = 8
level = 255 / (pow(2, bit_depth) - 1)
jumlah_level = pow(2, bit_depth)

print('Bit depth awal  :', bit_depth_awal, 'bit')
print('Bit depth tujuan:', bit_depth, 'bit')
print('Jumlah level    :', jumlah_level)

original = cv.imread('/content/drive/MyDrive/PCVK/Gambar/apel.jpg', cv.IMREAD_GRAYSCALE)
depth_image = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
    for x in range(original.shape[1]):
        depth_image[y, x] = round(original[y, x] / level) * level

final_frame = cv.hconcat((original, depth_image))
cv2_imshow(final_frame)
```

```python
# Nomor 8
# Average Denoising

original = cv.imread('/content/drive/MyDrive/PCVK/Gambar/galaxy.jpg')
cv_img = []
np.random.seed(42)
for i in range(100):
    noise = np.random.normal(0, 25, original.shape)
    noisy_image = original.astype(np.float32) + noise
    noisy_image = np.clip(noisy_image, 0, 255)
    noisy_image = np.uint8(noisy_image)
    cv_img.append(noisy_image)

def psnr(original, hasil):
    mse = np.mean((original.astype(np.float32) - hasil.astype(np.float32)) ** 2)
    if mse == 0:
        return float('inf')
    return 10 * math.log10((255 ** 2) / mse)
jumlah_citra = [10, 20, 40, 80, 100]

for jumlah in jumlah_citra:
    images = cv_img[:jumlah]
    average_image = np.mean(images, axis=0)
    average_image = np.uint8(average_image)
    nilai_psnr = psnr(original, average_image)
    print('Jumlah Citra:', jumlah)
    print('Nilai PSNR:', round(nilai_psnr, 2), 'dB')
    cv2_imshow(average_image)
```

```python
# Nomor 9
# Image Masking

original = cv.imread('/content/drive/MyDrive/PCVK/Gambar/couple.png')
h, w = original.shape[:2]
mask1 = np.zeros((h, w), dtype=np.uint8)
mask2 = np.zeros((h, w), dtype=np.uint8)
cv.circle(mask1, (int(w * 0.30), int(h * 0.18)), int(h * 0.16), 255, -1)
cv.circle(mask2, (int(w * 0.70), int(h * 0.20)), int(h * 0.16), 255, -1)
mask = cv.bitwise_or(mask1, mask2)

gabungan = np.hstack((original, cv.cvtColor(mask, cv.COLOR_GRAY2BGR)))
cv2_imshow(gabungan)

not_mask = cv.bitwise_not(mask)
or_mask = cv.bitwise_or(mask1, mask2)
and_mask = cv.bitwise_and(mask1, mask2)
nand_mask = cv.bitwise_not(and_mask)
xor_mask = cv.bitwise_xor(mask1, mask2)

not_image = cv.bitwise_and(original, original, mask=not_mask)
or_image = cv.bitwise_and(original, original, mask=or_mask)
and_image = cv.bitwise_and(original, original, mask=and_mask)
nand_image = cv.bitwise_and(original, original, mask=nand_mask)
xor_image = cv.bitwise_and(original, original, mask=xor_mask)

print('NOT (Komplemen)')
cv2_imshow(not_image)
print('OR (Atau)')
cv2_imshow(or_image)
print('AND (Dan)')
cv2_imshow(and_image)
print('NAND (Not And)')
cv2_imshow(nand_image)
print('XOR (Exclusive Or)')
cv2_imshow(xor_image)
```
