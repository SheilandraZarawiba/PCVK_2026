```python
import cv2 as cv
from matplotlib import pyplot as plt
from skimage import filters
from skimage.morphology import skeletonize

# read image
img = cv.imread('/content/drive/MyDrive/PCVK/Gambar/lily.jpg',0)

binary = img > filters.threshold_triangle(img)

# true false to one
binary_cp = binary.copy()
binary_cp[binary_cp == True] = 1
binary_cp[binary_cp == False] = 0

# skeletonize image
skeleton = skeletonize(binary_cp)

# print images
plt.figure(figsize=(20,20))

plt.subplot(121),plt.imshow(img,cmap = 'gray')
plt.title('Original Image'), plt.xticks([]), plt.yticks([])

plt.subplot(122),plt.imshow(skeleton,cmap = 'gray')
plt.title('skeleton Image'), plt.xticks([]), plt.yticks([])

plt.savefig("binary and bin_skeleton.png")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Membaca citra KTP
path = '/content/drive/MyDrive/PCVK/Gambar/ktp fake.jpeg'
img = cv2.imread(path)

if img is None:
    raise FileNotFoundError(f'Gambar tidak ditemukan: {path}')

# Mengubah citra menjadi grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Mengurangi noise dan meningkatkan kontras
gray_blur = cv2.GaussianBlur(gray, (3, 3), 0)

# Mengubah citra menjadi citra biner menggunakan Otsu
ret, thresh = cv2.threshold(
    gray_blur, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

# Menentukan structuring element
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3, 3))

# Operasi erosi
erosion = cv2.erode(thresh, kernel, iterations=1)

# Operasi dilasi
dilation = cv2.dilate(thresh, kernel, iterations=1)

# Operasi opening
opening = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)

# Operasi closing
closing = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)

# Menampilkan hasil
plt.figure(figsize=(16, 10))

plt.subplot(231)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Citra KTP Awal')
plt.axis('off')

plt.subplot(232)
plt.imshow(gray, cmap='gray')
plt.title('Citra Grayscale')
plt.axis('off')

plt.subplot(233)
plt.imshow(thresh, cmap='gray')
plt.title('Citra Biner')
plt.axis('off')

plt.subplot(234)
plt.imshow(erosion, cmap='gray')
plt.title('Hasil Erosi')
plt.axis('off')

plt.subplot(235)
plt.imshow(dilation, cmap='gray')
plt.title('Hasil Dilasi')
plt.axis('off')

plt.subplot(236)
plt.imshow(opening, cmap='gray')
plt.title('Hasil Opening')
plt.axis('off')

plt.tight_layout()
plt.show()

# Menampilkan hasil closing secara terpisah
plt.figure(figsize=(12, 5))

plt.subplot(121)
plt.imshow(thresh, cmap='gray')
plt.title('Citra Biner Awal')
plt.axis('off')

plt.subplot(122)
plt.imshow(closing, cmap='gray')
plt.title('Hasil Closing')
plt.axis('off')

plt.tight_layout()
plt.show()
```
