```python

import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg","https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg"]
# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
  image = io.imread(url)                          #read image
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize image to half size
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)  #convert color to RGB
  final_frame = cv.hconcat((image, image_2))      #concatenate image
  cv2_imshow(final_frame)
  print("\n")
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi, " x ", lebar)
cv2_imshow(image_2)

resolusi image: tinggi x lebar =  286  x  400
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# membuat garis horizontal ditengah image
for y in range(lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pyplot as plt
```

```python
urls = [
    "https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
    "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg"
]

url = urls[1]

image = io.imread(url)

image = cv.resize(image, (0, 0), fx=0.5, fy=0.5)

image = cv.cvtColor(image, cv.COLOR_BGR2RGB)

cv2_imshow(image)
```

```python
#Nomor 1
tinggi = image.shape[0]
lebar = image.shape[1]

print("Tinggi :", tinggi)
print("Lebar  :", lebar)
image_horizontal = image.copy()

for y in range(lebar):
    image_horizontal[int(tinggi / 2), y] = [255, 255, 255]

cv2_imshow(image_horizontal)
```

```python
#Nomor 2 A
image_vertical = image.copy()

for x in range(tinggi):
    image_vertical[x, int(lebar / 2)] = [255, 255, 255]

cv2_imshow(image_vertical)
```

```python
#Nomor 2 B
image_diagonal = image.copy()

for i in range(min(tinggi, lebar)):
    image_diagonal[i, i] = [255, 255, 255]

for i in range(min(tinggi, lebar)):
    image_diagonal[i, lebar - 1 - i] = [255, 255, 255]

cv2_imshow(image_diagonal)
```

```python
#Nomor 3
image_horizontal_panjang = image.copy()

y = int(tinggi / 2)

x_awal = int((lebar - 200) / 2)
x_akhir = x_awal + 200

image_horizontal_panjang[y, x_awal:x_akhir] = [255, 255, 255]

cv2_imshow(image_horizontal_panjang)
```

```python
#Nomor 4
image_kotak = image.copy()

image_kotak[100:200, 200:300] = [255, 255, 255]

cv2_imshow(image_kotak)
```

```python
black_img = np.zeros(shape=(512,512,3),dtype=np.int16)

plt.imshow(black_img)
```

```python
cv.rectangle(black_img,pt1=(200,200),pt2=(300,300),color=(0,0,255),thickness=15)

plt.imshow(black_img)
```

```python
cv.line(black_img,pt1=(0,0),pt2=(512,512),color=(255,255,255),thickness=5)

plt.imshow(black_img)
```
